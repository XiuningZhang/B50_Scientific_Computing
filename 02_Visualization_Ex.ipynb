## 1) Plotting curves
a) Draw a spiral.

```python
t = np.linspace(0, 10 * np.pi, 1000)
a = 0.05
x = a * t * np.cos(t)
y = a * t * np.sin(t)
plt.plot(x, y, color='gray', linewidth=2)
plt.axis('equal')
plt.axis('off')
plt.show()
```

b) Draw a "$\infty$" shape.

```python
t = np.linspace(0, 2 * np.pi, 1000)
a = 1
x = a * np.sin(t)
y = a * np.sin(2 * t)
plt.plot(x, y, color='cyan', linewidth=2)
plt.axis('equal')
# plt.axis('off')
plt.show()
```

```python
t = np.linspace(0, 2 * np.pi, 1000)
a = 1
x = a * np.cos(t) / (1 + np.sin(t)**2)
y = a * np.sin(t) * np.cos(t) / (1 + np.sin(t)**2)
plt.plot(x, y, color='royalblue', linewidth=2)
plt.axis('equal')
# plt.axis('off')
plt.show()
```

c) Draw a "flower-like" shape.

```python
t = np.linspace(0, 2 * np.pi, 1000)
a = 10
x = np.cos(t) + np.sin((a-1) * t)
y = np.sin(t) + np.cos((a-1) * t)
plt.plot(x, y, color='red', linewidth=2)
plt.axis('equal')
# plt.axis('off')
plt.show()
```

b) Create a matrix of pair-wise scatter plots like this:  
![pairs](https://github.com/oist/iSciComp/blob/master/figures/pairs.png?raw=1)
You can use `plt.tight_layout()` to adjust the space between subplots.

```python
plt.figure(figsize=(12, 12))
d = 4
for i in range(1, d):
    for j in range(d-1):
        if j < i:
            plt.subplot(4, 4, i*4+j+1)
            plt.scatter(X[:, i], X[:, j], c=Y)
            plt.xlabel("X{0}".format(i))
            plt.ylabel("X{0}".format(j))
plt.tight_layout()
```

```python
fig, axs = plt.subplots(3, 3, figsize=(12, 12))
for i in range(3):
    for j in range(3):
        if i <= j:
          axs[j, i].scatter(X[:, i], X[:, j+1], c=Y)
          axs[j, i].set_xlabel("X"+str(i)+":"+str(iris.feature_names[i]))
          axs[j, i].set_ylabel("X"+str(j+1)+":"+iris.feature_names[j+1])
        else:
          axs[j, i].axis('off')
plt.tight_layout()
```

## 3) Surface plots
a) Draw a wavy surface (not just a sine curve extended in the 3rd dimension).

```python
%matplotlib widget
```

```python
%matplotlib inline
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)/R
ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
```

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X)*np.sin(Y)
ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.xlabel('x'); plt.ylabel('y'); ax.set_zlabel('z');
```

b) Draw the surface of a (half) cylinder.  
Note that the mesh grid does not need to be square.

A half cylinder (0 <= theta <= pi), using a square mesh grid:

```python
x = np.linspace(-10, 10, 1000)
y = np.linspace(-10, 10, 1000)
X, Y = np.meshgrid(x, y)
R = 10
Z = np.sqrt(R**2 - X**2)
ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_zlim(-10, 10)
```

A full cylinder (0 <= theta < 2pi), using a cylindrical mesh grid:

```python
theta = np.linspace(0, 2 * np.pi, 100)
z = np.linspace(-5, 5, 100)
THETA, Z = np.meshgrid(theta, z)
R = 10
X = R * np.cos(THETA)
Y = R * np.sin(THETA)
ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
```

c) Draw the surface of a sphere.

```python
theta = np.linspace(0, 2 * np.pi, 100)
phi = np.linspace(-np.pi/2, np.pi/2, 100)
THETA, PHI = np.meshgrid(theta, phi)
R = 10
X = R * np.cos(PHI) * np.cos(THETA) 
Y = R * np.cos(PHI) * np.sin(THETA)
Z = R * np.sin(PHI)
ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
```
